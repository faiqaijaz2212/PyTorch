# Creating Tensors

```python
# using empty
a=torch.empty(2, 3) #allocates memory but doesn't initialize values. shows random values from previous memory.
print(a)
```

```python
# check type
print(type(a))
```

```python
#using zeros
b = torch.zeros(2, 3)
print(b) # all values are zero
```

```python
# using ones
c = torch.ones(2, 3)
print(c) # all values are one
```

```python
# using rand
d = torch.rand(2, 3)
print(d) # random values between 0 and 1
```

```python
# use of seed
d = torch.rand(2, 3)
print(d)
```

```python
# manual seed
torch.manual_seed(100)
f = torch.rand(2, 3)
print(f)
```

```python
torch.manual_seed(100)
f = torch.rand(2, 3)
print(f) # same values as before
```

```python
# using tensor
torch.tensor([[1, 2, 3],[4, 5, 6]])
```

```python
# other ways to create tensors

# arange
t = torch.arange(0, 10, 2) # creates a 5 element tensor with values from 0 to 10 having step size 2
print(t)

#using linspace
t = torch.linspace(0, 10, 5) # creates a 5 element tensor with values from 0 to 10 regularly spaced
print(t)

# using eye
t = torch.eye(3) # creates an identity matrix
print(t)

# using full
t = torch.full((2, 3), 5) # creates a 2x3 tensor filled with 5
print(t)
```

```python
b+5
```

```python
import time

size=10000

matrix_cpu1=torch.rand(size, size)
matrix_cpu2=torch.rand(size, size)

start_time=time.time()
result_cpu=matrix_cpu1 @ matrix_cpu2 #matrix multiplication
end_time=time.time()
cpu_time = end_time-start_time
print(f"CPU time: {cpu_time:.4f} seconds")

matrix_gpu1=matrix_cpu1.to("cuda")
matrix_gpu2=matrix_cpu2.to("cuda")

start_time=time.time()
result_gpu=matrix_gpu1 @ matrix_gpu2 #matrix multiplication
end_time=time.time()
gpu_time = end_time-start_time
print(f"GPU time: {gpu_time:.4f} seconds")
print("\nSpeedup (CPU/GPU):", cpu_time / gpu_time)
```
